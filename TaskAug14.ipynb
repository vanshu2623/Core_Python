### Examples of String with operators

```python
first = "Hello"
second = "Python"

result = first + " " + second
# result = first - " " - second   TypeError: unsupported operand type(s) for -: 'str' and 'str'

print(result)
print(first * 2)
```

```python
text = "Python "

print(text * 3)
```

### Examples in int 

```python
a = "Python"
b = "3"

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
print(a * 3)
# print(a * 2.5)   TypeError: can't multiply sequence by non-int of type 'float' 
```

```python
s1 = "Two"
num1 = 2
print(s1 == num1)
```

### Examples in Boolean

```python
a = True
b = False

print(a and b)
print(a or b)
print(not a)
print(not b)
print(a == b)
print(a != b)

```

### Examples in Float

```python
p = 10.5
q = 2.5

print(p + q)
print(p - q)
print(p * q)
print(p / q)
print(p // q)
print(p % q)
print(p ** q)

print(p > q)
print(p < q)
```
