# Operators
### Relational / Comparison Operators:- 
### Relational operators always returns boolean value (True / False)
### (>,<,>=,<=>)
### Equality operator:- (== , !=)
### == (Equals to Equals) , != (Not Equals to)

```python
marks1 = 75
marks2 = 80

print(marks1 > marks2)
print(marks1 < marks2)
print(marks1 == marks2)
print(marks1 != marks2)
```

```python
marks1 = 55
marks2 = 55
result1 = marks1 == marks2
result2 = marks1 != marks2
print(result1 , result2)
```

## Conditional Statements:-
### if :->
### To check half condition
### if - else :-
### We use if - else to check one complete condition
### sytnax:->      if condition : 
###                     if block code
###                else :
###                     else block code

```python
marks = 75

if marks >= 90:
    print("A")
elif marks >= 60:
    print("B")
else:
    print("C")
```

```python
age = 18

if age >= 18:
    print("Eligible for voting")
else:
    print("Not eligible for voting")
```

## Logical Operators:-
### for logical operations on data we use Logical operators
## and:- 
### If all the operands are True then and then only output is True otherwise false.

```python
username = "admin"
password = "1234"

if username == "admin" or password == "1234":
    print("Login successful")
else:
    print("Invalid username or password")
```

```python
a = 10
b = 25
c = 15

if a > b and a > c:
    print("A is largest")
elif b > a and b > c:
    print("B is largest")
else:
    print("C is largest")  
```

## or:- If any operand is True then op is True.
### If all the operands are False then and then only output is false.

```python
day = "Sunday"

if day == "Saturday" or day == "Sunday":
    print("It's weekend")
else:
    print("It's a working day")
```

## not(Inversion operator)--> 
### The not operator reverses the result of a condition.
### not True → False
### not False → True

```python
day = "Sunday"

if not(day == "Saturday" or day == "Sunday"):
    print("It's weekend")
else:
    print("It's a working day")
```
