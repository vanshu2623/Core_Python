## Composite/Compound Assignment:- 
### A composite assignment operator combines an arithmetic/bitwise operation with the assignment = operator.
### (+= , -= , *= , /= , //= , **=)

```python
x = 5

x += 10
x *= 2
x **= 2

print(x)
```

```python
marks = 60

marks += 10
marks *= 2
marks -= 20

print(marks)
```

## Bitwise Operator:-
### Bitwise operators work on the binary bits (0 and 1) of integers.
### (& , | , ^ , ~ , >>, <<, >>>)
### Integer:- We can represent integers by 4 ways.
## 1.Decimal:- Base:10 ,Digits:0–9 ,No prefix.

```python
a = 10
print(a)
print(type(a))
```

## 2.Binary:- Base:2, Digits:0,1 ,Prefix: 0b or 0B

```python
a1 = 0b1010
a2 = 0B1010
print(a1)
print(a2)
print(type(a1))
```

## 3.Octal:- Base: 8, Digits: 0–7, Prefix: 0o or 0O

```python
a = 0o1010
b = 0o67
print(a)
print(b)
```

## 4.Hexadecimal:- Base: 16, Digits: 0–9, a-f,A–F, Prefix: 0x or 0X

```python
var = 0xface
a = 0xA
print(var)
print(a)
```

### Examples of bitwise operators using integers.

#### 1. Bitwise AND (&)

```python
a = 5
b = 3

print(a & b)
```

#### 2. Bitwise OR ( | )

```python
a = 5
b = 3

print(a | b)
```

#### 3. Bitwise XOR (^)

```python
a = 5
b = 3

print(a ^ b)
```

#### 4. Bitwise NOT (~)

```python
a = 5

print(~a)
```

#### 5. Right Shift (>>)

```python
a = 5
b = 8
print(a >> 2)
print(b >> 2)
```

#### 6. Left Shift (<<)

```python
a = 5
b = 1
print(a << 2)
print(b << 2)
```
