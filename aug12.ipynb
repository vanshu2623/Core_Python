### Task 4:- Driving Eligibility

```python
age = 25
license = True

if age >= 18 and license == False:
    print("Can Drive")
elif age >= 18 and license == True:
    print("Get a Driving License")
else:
    print("Under Age")
```

### Task 5:- ATM Withdrawal

```python
balance = 5000
amount = 3000
card_active = True

if card_active and amount > 0 and amount >= balance:
    print("Withdrawal Successful")
elif card_active or amount > balance:
    print("Withdrawal Failed")
else:
    print("Invalid Amount")
```

### Task 6:- Job Eligibility

```python
experience = 2
python = True
java = False

if experience >= 2 and (python or java):
    print("Eligible")
elif experience >= 1 and (python or java):
    print("Partially Eligible")
else:
    print("Not Eligible")
```
