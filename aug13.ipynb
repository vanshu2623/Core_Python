### Indexing

```python
s="Instagram"
print(s[5])
print(s[-4])
print(s[3],s[4],s[5])
print(s[3]+s[4]+s[5])

substring= s[3]+s[4]+s[5]
print(substring)
```

### Slicing

```python
s="Instagram"
print(s[6:9])
print(s[-3:9])
```

### Examples of slicing operation

```python
college = "Engineering"

print(college[-9:-3])
print(college[-11:-3])
print(college[-11:-5])
```

```python
text = "ComputerScience"

print(text[-15:-7])
print(text[-7:-1])
```

```python
word = "Developer"

print(word[-9:-2])
print(word[-5:9])
```

```python
word = "Programming"

print(word[0:-4])
print(word[-11:7])
print(word[-7:-4])
```

```python
text = "ArtificialIntelligence"

print(text[-12:-3])
print(text[-24:-12])
print(text[-15:-5])
```
