### Examples of String methods

### 1. replace() :- Replaces one string with another.

```python
text = "I like Java"
print(text.replace("Java", "Python"))
```

### 2. split() :- Splits a string into a list.

```python
text = "Python Java C"
print(text.split())
print("-".join(text))
```

### 3. join() :- Joins elements into a string.

```python
words = ["Python", "Java", "C"]
print("-".join(words))
```

### 4. strip() :- Removes spaces from the beginning and end.

```python
text = "  Python  "
print(text.strip())
```

### 5. find() :- Returns the index of the first occurrence.
### 6. capitalize :- Converts the first character to uppercase.

```python
text = "Hello Python"
print(text.find("Python"))
name = "python programming"
print(name.capitalize())
```

### 7. swapcase() :- Converts uppercase to lowercase and lowercase to uppercase.
### 8. isspace() :- Checks whether all characters are spaces/whitespace.

```python
text = "PyThOn"
print(text.swapcase())
text = "   "
print(text.isspace())
```

### 9. casefold() :- Converts a string to lowercase for case-insensitive comparison.
### 10. partition() :- Divides a string into three parts based on the first occurrence of a separator.
### 11. center() :- Places a string in the center within a specified width.

```python
text = "HELLO"
print(text.casefold())
text = "Python-Java"
print(text.partition("-"))
text = "Python"
print(text.center(10, "-"))
```
