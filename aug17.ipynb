### Examples of String methods

### 1. replace() :- Replaces one string with another.

```python
text = "I like Java"
print(text.replace("Java", "Python"))
```

### 2. split() :- Splits a string into a list.

```python
text = "Python Java C"
print(text.split())
print("-".join(text))
```

### 3. join() :- Joins elements into a string.

```python
words = ["Python", "Java", "C"]
print("-".join(words))
```

### 4. strip() :- Removes spaces from the beginning and end.

```python
text = "  Python  "
print(text.strip())
```

### 5. find() :- Returns the index of the first occurrence.
### 6. capitalize :- Converts the first character to uppercase.

```python
text = "Hello Python"
print(text.find("Python"))
name = "python programming"
print(name.capitalize())
```

### 7. swapcase() :- Converts uppercase to lowercase and lowercase to uppercase.
### 8. isspace() :- Checks whether all characters are spaces/whitespace.

```python
text = "PyThOn"
print(text.swapcase())
text = "   "
print(text.isspace())
```

### 9. casefold() :- Converts a string to lowercase for case-insensitive comparison.
### 10. partition() :- Divides a string into three parts based on the first occurrence of a separator.
### 11. center() :- Places a string in the center within a specified width.

```python
text = "HELLO"
print(text.casefold())
text = "Python-Java"
print(text.partition("-"))
text = "Python"
print(text.center(10, "-"))
```

### 12. count() :- Counts how many times a character or substring occurs.
### 13. encode() :- Converts the string into encoded bytes.
### 14. find() :- Finds the first position of a value.

```python
text = "banana"
print(text.count("a"))
print(text.encode())
print(text.find("a"))
```

### 15. startwith() :- Checks whether a string starts with a particular value.
### 16. endswith() :- Checks whether a string ends with a particular value.

```python
text = "Python.py"
print(text.startswith("py"))
print(text.endswith(".py"))
```

### 17. isalnum() :- Checks whether all characters are letters or numbers.
### 18. isalpha() :- Checks whether all characters are alphabetic.
### 19. isascii() :- Checks whether all characters are ASCII.
### 20. isdigit() :- Checks whether all characters are digits.

```python
text = "Hello123"
print(text.isalnum())
print(text.isalpha())
print(text.isdigit())
print(text.isascii())
```

### 21. isidentifier() :- Checks whether a string can be used as a valid Python identifier.
### 22. isprintable() :- Checks whether all characters are printable.

```python
text = "student_name"
print(text.isidentifier())
print(text.isprintable())
```

### 23. rsplit() :- Splits the string starting from the right side.
### 24. zfill() :- Adds zeros at the beginning.

```python
text = "A-B-C"
print(text.rsplit("-", 1))

number = "25"
print(number.zfill(5))
```
