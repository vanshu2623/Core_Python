### Make below code dynamic,
### Ask user on console using input function which character frequency you want
### and display free of that char

```python
s = "Programming"
c = input("Enter a char: ")
count = 0
for char in s:
    if char.lower() == c.lower():
        count += 1

print("Frequency of char", c, "=", count)      
```

```python
s = input("Enter a String:")
ch = input("Enter a character:")
count = 0
for char in s:
    if char.lower() == ch.lower():
        count += 1

print("Frequency of ch", ch, "=", count)
```

```python
s = input("Enter a String:")
count = 0
for char in s:
    if char == " ":
        count +=1

print(s)        
print("Total white spaces","=",count)        
```

```python
original_string = input("Enter a String:")
reverse_string = original_string[::-1]

if original_string.lower() == reverse_string.lower():
    print(original_string, "is a Palindrome") 
else:
    print(original_string, "is not a palindrome")           
```

### Examples of for loop

```python
s = input("Enter a String: ")
count = 0
for char in s:
    if char.lower() in "aeiou":
        count += 1

print("Total Vowels =", count)
```

```python
s = input("Enter a String: ")
ch = input("Enter a character: ")

position = -1

for i in range(len(s)):
    if s[i].lower() == ch.lower():
        position = i
        break

print("First occurrence =", position)
```

```python
s = input("Enter a String: ")
word = input("Enter a word: ")

if s.lower().startswith(word.lower()):
    print("Starts with", word)
else:
    print("Does not start with", word)
```

```python
s = input("Enter a String: ")
word = input("Enter a word: ")

if s.lower().endswith(word.lower()):
    print("Ends with", word)
else:
    print("Does not end with", word)
```

```python
s = input("Enter a String: ")
if s.isalnum():
    print("Contains only alphabets and numbers")
else:
    print("Contains special characters")
```

```python
s = input("Enter a String: ")
result = ""
for char in s:
    if char not in result:
        result += char

print("After removing duplicates =", result)
```

```python
s = input("Enter a String: ")
words = s.split()
longest = ""

for word in words:
    if len(word) > len(longest):
        longest = word

print("Longest word =", longest)
```

```python
n = int(input("Enter a number: "))

for i in range(1, n + 1):
    if i % 2 == 0:
        print(i, "Even")
    else:
        print(i, "Odd")
```
