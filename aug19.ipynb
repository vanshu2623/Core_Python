### Examples of List Methods

### 1.extend() :- Adds multiple elements to the end of a list.
### 2.remove() :- Removes the occurrence of a specified value.
### 3.pop() :- Removes an element using its index and returns that element.

```python
numbers = [10, 20, 30]
numbers.extend([40, 50, 60])
numbers.remove(30)
numbers.pop(0)
print(numbers)
```

### 4.index() :- Returns the index of the first occurrence of a value.
### 5.append() :- Counts how many times a value occurs.

```python
names = ["Amit", "Riya", "Rahul"]
names.append("Komal")
print(names.index("Riya"))
print(names)
```

### 6.copy() :- Creates a copy of the list.
### 7.reverse() :- Reverses the current order.
### 8.sort() :- Sorts the list in descending order.

```python
numbers = [10, 20, 20, 30]
new_numbers = numbers.copy()
numbers.reverse()
print(new_numbers)
print(numbers)
numbers.sort(reverse = True)
print(numbers)
```

### 9.count() :- Counts how many times a value occurs.
### 10.insert() :- Adds an element at a specific position.
### 11.clear() :- Removes all elements from the list. 

```python
numbers = [10, 20, 20, 40]
print(numbers.count(20))
numbers.insert(2, 30)
print(numbers)
numbers.clear()
print(numbers)
```
