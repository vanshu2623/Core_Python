# Operators in Python:- 
### Operators are used to perform operations on variables and values.
## 1.Arithmetic Operators ->
### Arithmetic operators are used to perform mathematical calculations on numbers in Python.

Addition,Subtraction,Multiplication

```python
a=6
b=2
print(a+b)
print(a-b)
print(a*b)
```

Division,Floor Division,Modulus,Power

```python
a=16.5
b=12
print(a/b)
print(a//b)
```

```python
a=27
b=2
print(a%b)
print(a**b)
```
